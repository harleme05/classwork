```python
import pandas as pd
import matplotlib.pyplot as plt
# %conda install seaborn
import seaborn as sns
from scipy import stats

```

```python
# %conda update -n base -c defaults conda
```

```python
url = "data/soccer-players.csv"
df = pd.read_csv(url)
df.head()
```

```python
# hist in plt
plt.figure(figsize=(5,3))
hts = df["height_in_cm"]
bins = round(hts.count()**(1/3))
plt.hist(x = hts, bins = bins, edgecolor = "black")
plt.xlabel("Height_in_cm")
plt.ylabel("Frq")
plt.title("Histogram of heights(cm) of soccer players")
plt.grid(True)
sk = stats.skew(hts, nan_policy="omit")
kurt = stats.kurtosis(hts, nan_policy="omit")
text = f"Skew:{sk:.3f}\nKurtosis:{kurt:.3f}"
 # plt.text(x=110, y=250, s=text, ha="center", va = "center", fontsize = 12, color = "orange")
plt.text(
    x=0.2, 
    y=0.90, 
    s=text, 
   # ha="center", 
    va = "center", 
    fontsize = 12, 
    color = "orange",
    transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color = "r", linestyle = "--", label = "Mean")
plt.axvline(x=hts.median(), color = "b",linestyle = "--", label="Median")
plt.legend()
```

```python
# hist in sns
df1 = df.loc[df["height_in_cm"]>150]
plt.figure(figsize=(16,6))
hts = df1["height_in_cm"]
sns.set_theme(style="darkgrid")
hist = sns.histplot(data=df1, x="height_in_cm", kde=True)
plt.xlabel("height_in_cm")
sk = stats.skew(hts, nan_policy="omit")
kurt = stats.kurtosis(hts, nan_policy="omit")
text = f"Skew:{sk:.3f}\nKurtosis:{kurt:.3f}"
 # plt.text(x=110, y=250, s=text, ha="center", va = "center", fontsize = 12, color = "orange")
plt.text(
    x=0.2, 
    y=0.90, 
    s=text, 
   # ha="center", 
    va = "center", 
    fontsize = 12, 
    color = "orange",
    transform=plt.gca().transAxes)
plt.axvline(x=hts.mean(), color = "r", linestyle = "--", label = "Mean")
plt.axvline(x=hts.median(), color = "b",linestyle = "--", label="Median")
plt.legend()

```
