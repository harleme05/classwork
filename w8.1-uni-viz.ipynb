```python
url = "data/soccer-players.csv"
df = pd.read_csv(url)
df.head()
```

```python
# hist in plt
plt.figure(figsize=(5,3))
hts = df["height_in_cm"]
plt.hist(x = hts, bins = 12, edgecolor = "black")
plt.xlabel("Height_in_cm")
plt.ylabel("Frq")
plt.title("Histogram of heights(cm) of soccer players")
plt.xlim(0,210)
plt.ylim(0,500)
plt.grid(True)
text = "skew : 0.12"
 # plt.text(x=110, y=250, s=text, ha="center", va = "center", fontsize = 12, color = "orange")
plt.text(
    x=0.1, 
    y=0.95, 
    s=text, 
   # ha="center", 
    va = "center", 
    fontsize = 12, 
    color = "orange",
    transform=plt.gca().transAxes)
plt.axvline(x=105, color = "r", linestyle = "--", label = "Mean")
plt.axhline(y=250, color = "b",linestyle = "--", label="Cenetr")
plt.show()
```
