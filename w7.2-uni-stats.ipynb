```python
from functions.eda import eda
import pandas as pd
df = pd.read_csv("data/soccer-players.csv")
eda(df)
```

```python
# date of birth and height in cm 
from scipy import stats
heights = df["height_in_cm"]
sk = stats.skew(heights, nan_policy="omit")
kurt = stats.kurtosis(heights.dropna())
print(f"skew = {sk:.3f} ; kurt = {kurt:.3f}")
```

```python
# IQR for outliers

Q1 = heights.quantile(.25)
Q3 = heights.quantile(.75)
IQR = Q3-Q1
IQR
```

```python
from functions.convert_height import convert_height
lower_bound = Q1 -1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = heights.loc[(heights<lower_bound)|(heights>upper_bound)]
for cm in outliers:
    print(convert_height(cm))
```
